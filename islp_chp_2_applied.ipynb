```python
boston = load_data('Boston')
boston.head()
```

```python
boston.describe()
```

```python
boston.info()
```

```python
from pandas.plotting import scatter_matrix

scatter_matrix(boston, figsize=(12,12));
```

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(3, 4, figsize=(12,10))

predictors = boston.columns.drop('crim')

for ax, var in zip(axes.ravel(), predictors):
    ax.scatter(boston[var], boston['crim'])
    ax.set_xlabel(var)
    ax.set_ylabel('crim')

plt.tight_layout()
```
